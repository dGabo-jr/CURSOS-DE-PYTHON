```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
nombre = ''
while nombre != 'Gabriel':
    nombre = input("ingrese la contraseña: ")
    if nombre != 'Gabriel':
        print('Ingrese  nuevamente la contraseña:  ')
    else:
        print(f"Su contraseña es correcta y es {nombre} ")
        break
print('Se ha salido del bucle')
```

#**TUPLAS**

Una tupla es una estructura de datos en la que se pueden agrupar varios elementos diferentes o del mismo tipo en un único objeto.

Las tuplas pueden ser de cualquier longitud y los elementos son ordenados de manera específica, lo que permite acceder a ellos mediante un índice numérico.

**las tuplas son inmutables, es decir, una vez creadas, no se pueden modificar**

#**DICCIONARIOS**

es una estructura de datos que almacena un conjunto de pares **clave-valor**. Cada elemento del diccionario se compone de una clave y un valor que se corresponden entre sí

###**por lo tanto la relevancia de este recae en la utilidad de asociar información específica a cada clave**
image.png

```python
futbolistas = dict()

futbolistas = {
    1 : "Casillas", 15 : "Ramos",
    3 : "Pique", 5 : "Puyol",
    11 : "Capdevila", 14 : "Xabi Alonso",
    16 : "Busquets", 8 : "Xavi Hernandez",
    18 : "Pedrito", 6 : "Iniesta",
    7 : "Villa"
}
```

```python
for k,v in futbolistas.items():
    print( "%s -> %s" %(k,v))
```
